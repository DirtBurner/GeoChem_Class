# Demonstration of $^{143}Nd$ ingrowth vs. $^{144}Nd$ decay in $\frac{^{143}Nd}{^{144}Nd}$ ratios

The time scale of changes in the ratio of $\frac{^{143}Nd}{^{144}Nd}$ is large. To demonstrate the following code cell imports packages to produce a graph of the evolution of this ratio and the separate changes in the numerator and denominator so that one can visualize the time scale and contribution of changes in this important ratio. 

```python
import numpy as np
import matplotlib.pyplot as plt

lambda_147 = 0.693/1.06e11
lambda_144 = 0.693/2.26e15
Sm147_Nd144_0 = [1, 2, 5, 10, 20, 30, 50]

#range of time representing age of the Earth:
t = np.linspace(1, 4.8e9, 1000)

fig, ax = plt.subplots(nrows=1, ncols=1)

Nd_143 = [1-np.exp(lambda_147*time) for time in t]
Nd_144 = [np.exp(-lambda_144*time) for time in t]
ratio_dict = {}
for rat in Sm147_Nd144_0:
    Nd_ratio = [rat/np.exp(lambda_144*time)-rat*np.exp(-lambda_147*time)/np.exp(lambda_144*time) for time in t]
    ratio_dict[rat] = Nd_ratio
    ax.plot(t, Nd_ratio, linewidth=2, label=r'($\frac{^{147}Sa}{^{144}Nd})_0$ = '+ str(rat), marker='')

ax.legend()
ax.set(xlabel='years', ylabel=r'$\frac{^{143}Nd}{^{144}Nd}$')


#ax.plot(t, Nd_143, linewidth=2, color='lightgreen', marker='')
#ax.plot(t, Nd_144, linewidth=2, color='lightblue', marker='')
#ax.plot(t, Nd_ratio, linewidth=2, color='khaki', marker='')

```
